```python
from dotenv import load_dotenv
from IPython.display import Markdown, display, update_display

load_dotenv()
from tech_explainer.cli import stream_explanation

def explain(question, model="llama3.2"):
    """Stream the explanation into the notebook as live-updating Markdown."""
    response = ""
    handle = display(Markdown(""), display_id=True)
    for piece in stream_explanation(question, model):
        response += piece
        update_display(Markdown(response), display_id=handle.display_id)
```

```python
question = """
Please explain what this code does and why:
a = 43336
b = 4
c = a*b
"""
explain(question)
```
